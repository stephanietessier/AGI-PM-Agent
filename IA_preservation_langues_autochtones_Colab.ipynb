## Streamlit App for Recommendations

To run this Streamlit application, you will need to save the code to a `.py` file (e.g., `app.py`) and then execute it from your terminal using `streamlit run app.py`.

```python
import streamlit as st
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def recommander(theme, df, top_n=3):
    textes = df['categorie'] + ' ' + df['traduction_fr'] + ' ' + df['note_culturelle']
    vectorizer = TfidfVectorizer()
    matrix = vectorizer.fit_transform(textes)
    query_vec = vectorizer.transform([theme])
    scores = cosine_similarity(query_vec, matrix).flatten()
    result = df.copy()
    result['score_pertinence'] = scores
    return result.sort_values('score_pertinence', ascending=False).head(top_n)

st.title('AI de Préservation des Langues Autochtones')
st.write('Entrez un thème pour obtenir des recommandations de mots et de concepts:')

theme_input = st.text_input('Thème de recherche', 'nature eau vie')

if theme_input:
    # Assuming 'lexique' is accessible from the global scope of the Colab kernel.
    # If not, it would also need to be defined within this cell.
    recommendations = recommander(theme_input, lexique)
    st.dataframe(recommendations)
else:
    st.write('Veuillez entrer un thème pour obtenir des recommandations.')
```
